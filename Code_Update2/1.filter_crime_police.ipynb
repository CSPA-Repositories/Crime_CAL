```python
from IPython.display import display, HTML
def display_scrollable(df, max_rows=200, max_cols=10):
    html = df.to_html(max_rows=max_rows, max_cols=max_cols)
    display(HTML(f'<div style="height:400px;overflow:auto">{html}</div>'))
```

```python
display_scrollable(pd.DataFrame({'case_type': crime_filtered['case_type'].value_counts()}))
```

```python
crime_filtered = crime_filtered[~crime_filtered['case_type'].str.contains('纠纷|争议|行政', na=False)]
```

```python
print(len(crime_filtered))
crime_filtered = crime_filtered[['case_number', 'case_type', 'city', 'latitude', 'longitude','incident_location',
       'incident_province', 'incident_city', 'incident_county',
       'incident_time', 'formatted_datetime']]
```

```python
crime_filtered = crime_filtered.dropna(subset=['latitude', 'longitude'])
crime_filtered = crime_filtered[pd.to_numeric(crime_filtered['latitude'], errors='coerce').notnull()]
crime_filtered = crime_filtered[pd.to_numeric(crime_filtered['longitude'], errors='coerce').notnull()]
crime_filtered = crime_filtered.replace(r'\s+', ' ', regex=True)
```

```python
crime_filtered.to_csv("", index=False)
```
