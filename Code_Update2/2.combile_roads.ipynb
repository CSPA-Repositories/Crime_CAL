```python
level1_road_path = ""
level2_road_path = ""
level3_road_path = ""
level4_road_path = ""
gaosu_road_path = ""
guodao_road_path = ""
shengdao_road_path = ""
xiandao_road_path = ""
xiangdao_road_path = ""
import geopandas as gpd
import pandas as pd
gdf1 = gpd.read_file(level1_road_path).to_crs(epsg=4326)
gdf1["road_type"] = "一级"
gdf2 = gpd.read_file(level2_road_path).to_crs(epsg=4326)
gdf2["road_type"] = "二级"
gdf3 = gpd.read_file(level3_road_path).to_crs(epsg=4326)
gdf3["road_type"] = "三级"
gdf4 = gpd.read_file(level4_road_path).to_crs(epsg=4326)
gdf4["road_type"] = "四级"
gdf5 = gpd.read_file(gaosu_road_path).to_crs(epsg=4326)
gdf5["road_type"] = "高速"
gdf6 = gpd.read_file(guodao_road_path).to_crs(epsg=4326)
gdf6["road_type"] = "国道"
gdf7 = gpd.read_file(shengdao_road_path).to_crs(epsg=4326)
gdf7["road_type"] = "省道"
gdf8 = gpd.read_file(xiandao_road_path).to_crs(epsg=4326)
gdf8["road_type"] = "县道"
gdf9 = gpd.read_file(xiangdao_road_path).to_crs(epsg=4326)
gdf9["road_type"] = "乡道"
roads_combined = gpd.GeoDataFrame(
    pd.concat([gdf1, gdf2, gdf3, gdf4, gdf5, gdf6, gdf7, gdf8, gdf9], ignore_index=True),
    crs='EPSG:4326'
)
roads_combined.to_file("")
```

```python
city_list_path = ""
city_list_df = pd.read_csv(city_list_path)
city_list = city_list_df['地区'].tolist()
city_list = [str(city)+'市' for city in city_list]
region_path = ""
region_gdf = gpd.read_file(region_path)
region_gdf = region_gdf[region_gdf['市'].isin(city_list)]
region_gdf.sindex
roads_combined.sindex
roads_clipped = gpd.clip(roads_combined, region_gdf)
roads_clipped = gpd.sjoin(
    roads_clipped,
    region_gdf[['geometry', '省', '市', '县', '乡']],
    how='left',
    predicate='intersects'
)
roads_clipped = roads_clipped.drop(columns=['index_right'])
```

```python
roads_clipped.to_file("")
```
